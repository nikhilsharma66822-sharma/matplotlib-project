```python
df2=pd.read_csv("/content/employee_salary_dataset.csv")
df2
```

```python
plt.plot(df2['EmployeeID'], df2['Monthly_Salary'], color = '#4B5320',linestyle='solid', linewidth = 1, marker='o')

plt.title('EmployeeID vs. Monthly_Salary ')
plt.xlabel('EmployeeID')
plt.ylabel('Monthly_Salary')
plt.legend(['EmployeeID'])

plt.grid(True)
```

```python
plt.plot(df2['Monthly_Salary'], df2['Experience_Years'], color = '#4B5320',linestyle='solid', linewidth = 1, marker='o')

plt.title('Experience_Years vs. Monthly_Salary')
plt.xlabel('Monthly_Salary')
plt.ylabel('Experience_Years')
plt.legend(['Monthly_Salary'])

plt.grid(True)
```

```python
plt.plot(df2['Department'], df2['Age'], color = '#4B5320',linestyle='solid', linewidth = 1, marker='o')
plt.plot(df2['Department'], df2['Experience_Years'], color = '#4B5320',linestyle='solid', linewidth = 1, marker='o')

plt.title('EmployeeID vs. Monthly_Salary ')
plt.xlabel('EmployeeID')
plt.ylabel('Monthly_Salary')
plt.legend(['Age','Experience_Years'])

plt.grid(True)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df2['Experience_Years'], df2['Monthly_Salary'], color='#4B5320', s=50, alpha=0.7, edgecolor='black')

plt.title('Experience Years vs. Monthly Salary', fontsize=14, fontweight='bold')
plt.xlabel('Years of Experience', fontsize=12)
plt.ylabel('Monthly Salary', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df2['Monthly_Salary'], bins=8, color='darkolivegreen', edgecolor='black', alpha=0.8)

plt.title('Distribution of Employee Salaries', fontsize=14, fontweight='bold')
plt.xlabel('Monthly Salary Range', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
dept_salary = df2.groupby('Department')['Monthly_Salary'].mean()
plt.figure(figsize=(8, 5))
plt.bar(dept_salary.index, dept_salary.values, color='forestgreen', edgecolor='black', alpha=0.8)

plt.title('Average Monthly Salary by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department', fontsize=12)
plt.ylabel('Average Monthly Salary', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
